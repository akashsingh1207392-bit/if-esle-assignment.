# 10. Write a for loop to find the largest number in a list.

```python
numbers = [12, 45, 7, 89, 34, 56]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print("Largest number =", largest)
```

# 13. Write a for loop to reverse a string.

```python
text = "Akash"
reverse_text = ""

for char in text:
    reverse_text = char + reverse_text

print("Reversed string:", reverse_text)
```

# 14. Write a for loop to count vowels in a string.

```python
text = "Akash singh"

count = 0

for ch in text.lower():
    if ch in "aeiou":
        count += 1

print("Number of vowels:", count)
```

# 15. Write a for loop to print numbers from 1 to 20 but stop when the number is 12 (use
break).

```python
for num in range(1, 21):
    if num == 12:
        break
    print(num)
```

# 16. Write a for loop to print numbers from 1 to 20 but skip multiples of 4 (use continue).

```python
for num in range(1, 21):
    if num % 4 == 0:
        continue
    print(num)
```

# 17. Write a for loop to search for a number in a list and stop when it is found (use break).

```python
numbers = [10, 25, 30, 45, 50]
search_num = 30

for num in numbers:
    if num == search_num:
        print("Number found:", num)
        break
```

# 18. Write a for loop to print characters of a string but skip vowels (use continue).

# 19. Write a for loop to print numbers from 1 to 50 and stop when a number divisible by 9
is found (use break).

```python
for num in range(1, 51):
    if num % 9 == 0:
        print("First number divisible by 9 found:", num)
        break
    print(num)
```
